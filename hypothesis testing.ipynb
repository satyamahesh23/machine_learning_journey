```python
import numpy as np
from statsmodels.stats.weightstats import ztest
```

```python
salary = np.array([
    51000,49500,52000,50500,50000,
    49800,51500,49000,50800,50200,
    50600,49900,50100,51200,50700,
    50300,49600,50900,51100,50400
])
print(salary)
```

```python
population_mean=50000
z_statistic, p_value=ztest(salary,value=population_mean
)
print("Z Statistic :", z_statistic)

print("P Value :", p_value)
```

```python
alpha=0.05
# Decision based on p-value
if p_value < alpha:
    print("Reject Null Hypothesis (H0)")
else:
    print("Fail to Reject Null Hypothesis (H0)")
```

```python
battery = np.array([
    11,12,10,13,11,
    12,11,10,13,12
])
average_mean=10

z,p=ztest(battery,value=average_mean)
print("z-test",z)
print("p-test",p)
```

```python
alpha=0.05

if (p / 2 < alpha) and (z > 0):
    print("Reject H₀")
else:
    print("Fail to Reject H₀")
```
