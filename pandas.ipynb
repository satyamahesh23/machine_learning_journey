```python
sets={ # create a dict — each key is a column name
    'name': ['Alice', 'Bob', 'Charlie'], # column 1 — names as strings
'age': [25, 30, 22], # column 2 — ages as integers
'score': [88.5, 92.0, 76.3] # column 3 — scores as floats
    
}
df=pd.DataFrame(sets)
print(df.shape)
print(df.dtypes)
print(df.head(2))
print(df.describe())# .describe() = count, mean, std, min, max — instant stats summary
```

```python
df = pd.DataFrame({ # create sample data with missing values
'age': [25, np.nan, 22, 35], # np.nan = 'Not a Number' — standard missing value marker
'score': [88.5, 92.0, np.nan, 76.3] # missing score for 3rd person
})
 
print(df.isnull()) # .isnull() = True/False table — True means that cell is missing
print(df.isnull().sum())
```

```python
df_1 = pd.DataFrame({ # sample student dataset
'city': ['Hyd','Hyd','Delhi','Delhi','Hyd'], # categorical column — the group we want to split by
'score': [88, 92, 76, 85, 70], # numeric column — we'll compute stats on this
'age': [25, 30, 22, 28, 26] # another numeric column
})
df_1.groupby('city')['score'].mean()
df_1.groupby('city')['score'].agg(['mean','max','count'])
```
