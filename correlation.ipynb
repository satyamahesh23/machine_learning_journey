```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# create student data
data = {
    "Hours":[1,2,3,4,5],
    "Marks":[20,40,60,80,100]
}

# convert to dataframe
df = pd.DataFrame(data)

# draw scatter plot
# each dot represents one student
sns.scatterplot(
    x="Hours",
    y="Marks",
    data=df
    
)

# display graph
plt.show()
```

```python
data = {
    "Hours":[1,2,3,4,5],
    "Marks":[20,40,60,80,100],
    "TV":[5,4,3,2,1]
}

# convert to dataframe
df = pd.DataFrame(data)
corr=df.corr()
```

```python
sns.heatmap(
    corr,
    annot=True
)
plt.show()
```

```python
import numpy as np

x = np.array([1,2,3,4,5])
y = np.array([2,4,6,8,10])

mean_x = np.mean(x)
mean_y = np.mean(y)

# Difference from mean
x_diff = x - mean_x
y_diff = y - mean_y

print(x_diff)
print(y_diff)
```
