# different ways of creating numpy

```python
import numpy as np
```

```python
var_1=np.array([1,2,3,4])
print(var_1)
```

```python
var_2=np.array([[1,2,3],[5,6,7]])
print(var_2)
```

```python
#array of zeros:
np.zeros((2,3))
```

```python
#array of ones:
np.ones((2,2),dtype="int")
```
