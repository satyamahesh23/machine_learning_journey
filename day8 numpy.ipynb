# mini project : student marks analyzer

```python
marks=np.array([78, 65, 90, 45, 88, 72, 95, 60])
print("students marks :",marks)
print('total marks of all students :',marks.sum())
print("average marks of students :",marks.mean())
print("average marks of students :",marks.max())
print("average marks of students :",marks.min())
avg=marks[marks>marks.mean()]
print(avg)
print('no of students passed: ',(50<marks))


```
