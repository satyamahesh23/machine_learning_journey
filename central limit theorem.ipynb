```python
import pandas as pd
import seaborn as sns
```

```python
df=pd.read_csv('employee_lyst1729240478122.csv')
df
```

```python
population=df['Base Salary']
```

```python
print('population size',len(population))
```

```python
print("poplution mean",population.mean())
```

```python
print('population std',population.std())
```

# Step 3: Take one random sample

```python
sample=population.sample(n=30)
print(sample.mean())
```

# Step 4: Apply Central Limit Theorem

```python
sample_means = []

for i in range(1000):
    sample = population.sample(n=30, replace=True)
    sample_means.append(sample.mean())
```

# Step 5: Observe the result

```python
import matplotlib.pyplot as plt
```

```python
plt.hist(sample_means,bins=30)
plt.xlabel("Sample Means")
plt.ylabel("Frequency")
plt.title("Central Limit Theorem")
plt.show()
```
