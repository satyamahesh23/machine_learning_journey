scipy---------> scientific python. it is modified version of python

SciPy (Scientific Python) is a library built on top of NumPy that provides tools for:

📈 Statistics
🔢 Mathematical calculations
📉 Optimization
📊 Probability distributions
🧮 Solving equations
🖼️ Signal and image processing

Think of it as:

NumPy → Arrays
Pandas → Tables
Matplotlib → Graphs
SciPy → Advanced Mathematics and Science

```python
from scipy import stats
```

```python
marks= [80, 90, 90, 95, 100]
print(stats.mode(marks))
```

```python
marks = [50,60,70,80,90]
print("means:",stats.tmean(marks))
print("median:",stats.median_abs_deviation(marks))
```

```python
from scipy.special import factorial
print(factorial(5))
```

```python
from scipy import stats

marks = [78,85,90,78,88,95,78]

print("Average:", stats.tmean(marks))

print("Most Common Mark:", stats.mode(marks))

print("Standard Deviation:", stats.tstd(marks))
```
