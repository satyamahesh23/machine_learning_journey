```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson
```

```python
#parameters
lamba_value=5
k=np.arange(0,15)#possible values for k or t


#poissson
pmf_value=poisson.pmf(k,lamba_value)
#plotting
plt.bar(k,pmf_value,color="blue")
plt.title('poisson distribution')
plt.xlabel("number of events")
plt.ylabel("probability")
plt.grid(axis="y")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# average height = 165 cm, spread (std dev) = 7 cm
mean = 165
std = 7

# generate smooth range of x values (heights from 140 to 190)
x = np.linspace(140, 190)

# calculate the PDF curve values for each height
pdf_values = norm.pdf(x, mean, std)

# plot the smooth curve
plt.plot(x, pdf_values, color="blue")

# shade the area under curve between 159 and 161 (the "range" we care about)
x_fill = np.linspace(159, 161, 100)
y_fill = norm.pdf(x_fill, mean, std)
plt.fill_between(x_fill, y_fill, color="orange", alpha=0.6, label="Probability (area) for 159-161 cm")

plt.title("Probability Density Function (PDF) - Height Example")
plt.xlabel("Height (cm)")
plt.ylabel("Density")
plt.legend()
plt.grid(axis="y")
```
