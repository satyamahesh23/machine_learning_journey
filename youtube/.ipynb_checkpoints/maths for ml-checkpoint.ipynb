```python
import numpy as np # # numpy
 
features = np.array([3, 1200, 2]) # # one house: bedrooms, sqft, bathrooms
weights  = np.array([5000, 100, 3000]) # # how much each feature 'contributes' to price
 # # weights are LEARNED during training — this is what the model figures out
    
dot=np.dot(features,weights)
print(dot)
```

```python
bias=1000
prediction=np.dot(features,weights)+bias  # # y = w·x + b — the most fundamental ML formula
print(prediction)
```

```python
# Manual version (what dot product does under the hood):
manual = features[0]*weights[0] + features[1]*weights[1] + features[2]*weights[2] # # multiply matching positions, then sum everything
print(manual == dot) # # True — np.dot just does this automatically, much faster
```
