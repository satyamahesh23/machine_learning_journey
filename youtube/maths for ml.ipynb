```python
import numpy as np # # numpy
 
features = np.array([3, 1200, 2]) # # one house: bedrooms, sqft, bathrooms
weights  = np.array([5000, 100, 3000]) # # how much each feature 'contributes' to price
 # # weights are LEARNED during training — this is what the model figures out
    
dot=np.dot(features,weights)
print(dot)
```

```python
bias=1000
prediction=np.dot(features,weights)+bias  # # y = w·x + b — the most fundamental ML formula
print(prediction)
```

```python
# Manual version (what dot product does under the hood):
manual = features[0]*weights[0] + features[1]*weights[1] + features[2]*weights[2] # # multiply matching positions, then sum everything
print(manual == dot) # # True — np.dot just does this automatically, much faster
```

```python
import numpy as np # # numpy
 
X = np.array([ # # dataset matrix: 3 houses x 3 features
  [3, 1200, 2], # # house 1
  [2,  900, 1], # # house 2
  [4, 1800, 3] # # house 3
])
 
weights=np.array([5000,100,3000])
predictions=X@weights
print(predictions)
```

```python
import numpy as np # # numpy
 
point = np.array([2, 3]) # # a single 2D data point: x=2, y=3
 
scale_matrix = np.array([ # # a transformation matrix that scales x by 2, y by 3
  [2, 0], # # row 1: scales x-axis by 2, no effect on y
  [0, 3] # # row 2: no effect on x, scales y-axis by 3
])
 
scaled_point = scale_matrix @ point # # matrix multiplication transforms the point
print(scaled_point) # # result: [4, 9] — x doubled, y tripled
```

```python
rotate_90=np.array([
    [0,-1],
    [1,0]
])
rotated_point=rotate_90@point
print(rotated_point)
```
