```python
import pandas as pd # # step 1: import pandas — we need it to create our dataset
from sklearn.preprocessing import MinMaxScaler # # step 2: import MinMaxScaler — it applies the normalization formula
 
# Create a raw dataset with very different scales
df = pd.DataFrame({ # # pd.DataFrame() converts a dictionary into a table
  'age':    [20, 35, 50, 28, 60], # # age column — range is 20 to 60 (small numbers)
  'salary': [10000, 50000, 120000, 35000, 200000] # # salary column — range is 10k to 2L (huge numbers)
})
 
scaler = MinMaxScaler() # # create scaler object — formula it uses: (x - min) / (max - min)
scaled_array = scaler.fit_transform(df) # # fit = learn min and max from data; transform = apply formula
 # # why fit_transform together? so we learn AND apply in one step
df_scaled = pd.DataFrame(scaled_array, columns=df.columns) # # wrap result back into DataFrame — keeps column names
 
print(df_scaled) # # every value is now between 0.0 and 1.0
print('Min values:\n', df_scaled.min()) # # should print 0.0 for each column (the smallest original value → 0)
print('Max values:\n', df_scaled.max()) # # should print 1.0 for each column (the largest original value → 1)
```
