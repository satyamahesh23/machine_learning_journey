```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# create values from -4 to +4
x = np.linspace(-4,4,100)

# create normal distribution curve
y = norm.pdf(x)

# plot graph
plt.plot(x,y)

# title
plt.title("Normal Distribution")

# display graph
plt.show()
```
