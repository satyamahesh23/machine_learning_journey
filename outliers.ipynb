# how to detect outliers

```python
import seaborn as sns
```

```python
sns.distplot(df['cgpa'])
```

```python
#finding the boundary
low=df['cgpa'].mean()-3*(df['cgpa'].std())
upp=df['cgpa'].mean()+3*(df['cgpa'].std())
```

```python
low
```

```python
upp
```

```python
df[df['cgpa']<low]
```

```python
df[df['cgpa']>upp]
```
