```python
import pandas as pd
```

```python
df=pd.read_csv('feature_engineering_practice_dataset.csv')
print(df)
```

```python
print(df.shape)
```

```python
print(df.isnull().sum())
```

```python
df["Python"]=df["Python"].fillna(df["Python"].mean())
print(df)
```

```python
df['SQL']=df['SQL'].fillna(df['SQL'].mean())
print(df)
```

```python
df['ML']=df['ML'].fillna(df['ML'].mean())
print(df)
```

```python
df['Total']=(df['Python']+df['SQL']+df['ML'])
print(df)
```

```python
df['Average']=df['Total']/3
print(df)
```

```python
df["JoinDate"] = pd.to_datetime(
    df["JoinDate"]
)
print(df)
```

```python
print(df.loc[df['Average'].idxmax()])
```

```python
print(df.loc[df['Attendance'].idxmax()])
```

```python
print([df['Average']>85])
```

```python
print(df.sort_values('SQL',ascending=False))
```

```python
def att(x):
    if x>=90:
        return 'excellent'
    elif x>80:
        return 'good'
    elif x>70:
        return "ok"
    else:
        return 'poor'
df['Attendance_Category']=(df['Attendance'].apply(att))
print(df)
```

```python
df['Performance']=(df['Average']*df['Attendance']/100)
print(df)
```

```python
df['scholarship']=((df['Average']>85)&(df['Attendance']>90).map({
    True:'Yes',
    False:'No'
}))
print(df)
```
